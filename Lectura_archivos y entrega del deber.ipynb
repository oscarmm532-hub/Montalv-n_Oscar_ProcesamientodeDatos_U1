```python
from pathlib import Path
import pandas as pd


def leer_archivo(ruta):
    """Lee archivos CSV, TXT o JSON."""
    ruta = Path(ruta)

    if not ruta.exists():
        raise FileNotFoundError(f"No existe el archivo: {ruta}")

    extension = ruta.suffix.lower()

    if extension == ".csv":
        return pd.read_csv(ruta, sep=None, engine="python")

    elif extension == ".txt":
        # Para TXT delimitados o archivos de texto libre.
        try:
            df = pd.read_csv(
                ruta, sep=None, engine="python", header=None
            )
            if df.shape[1] == 1:
                df.columns = ["texto"]
            else:
                df.columns = [
                    f"columna_{i+1}" for i in range(df.shape[1])
                ]
            return df
        except Exception:
            return pd.DataFrame({
                "texto": ruta.read_text(
                    encoding="utf-8",
                    errors="replace"
                ).splitlines()
            })

    elif extension == ".json":
        try:
            return pd.read_json(ruta)
        except ValueError:
            # Para JSON Lines: un objeto JSON por línea.
            return pd.read_json(ruta, lines=True)

    else:
        raise ValueError(
            "Formato no admitido. Use CSV, TXT o JSON."
        )


def ver_esquema(df):
    """Muestra el esquema y una vista previa."""
    print("Número de registros:", len(df))
    print("Número de columnas:", len(df.columns))

    esquema = pd.DataFrame({
        "Columna": df.columns,
        "Tipo": [str(df[c].dtype) for c in df.columns],
        "Valores nulos": [
            df[c].isna().sum() for c in df.columns
        ],
        "Valores distintos": [
            df[c].nunique(dropna=True) for c in df.columns
        ]
    })

    print("\nESQUEMA DEL ARCHIVO")
    print(esquema.to_string(index=False))

    print("\nPRIMERAS CINCO FILAS")
    print(df.head().to_string(index=False))

    # cambiar por la ruta de su archivo
ruta = "personas-100.csv"

try:
    datos = leer_archivo(ruta)
    ver_esquema(datos)

except Exception as error:
    print("Error:", error)


```

```python
import json
import pandas as pd

ruta_nba_nested = "nbaplayers-nested.json"

try:
    # Leemos el archivo JSON nativo
    with open(ruta_nba_nested, 'r', encoding='utf-8') as archivo:
        datos_crudos = json.load(archivo)

    # Extraemos la lista que está dentro de "data" y creamos el DataFrame
    df_nba_nested = pd.DataFrame(
        datos_crudos["data"],
        columns=["id", "player_name", "team_name"]
    )

    print("--- DATOS NBA NESTED ---")
    ver_esquema(df_nba_nested)
except Exception as error:
    print("Error:", error)
```

```python
import json
import pandas as pd
import re

ruta_pizzas = "pizzasmilan.json"

try:
    # 1. Leemos el archivo completo como texto plano
    with open(ruta_pizzas, 'r', encoding='utf-8') as archivo:
        contenido = archivo.read()

    # 2. Arreglamos el formato uniendo los bloques sueltos con una coma
    contenido_arreglado = re.sub(r'\}\s*\{', '},{', contenido)

    # 3. Lo encerramos entre corchetes para convertirlo en una lista JSON válida
    json_valido = "[" + contenido_arreglado + "]"

    # 4. Cargamos los datos y creamos el DataFrame
    datos_crudos = json.loads(json_valido)
    df_pizzas = pd.DataFrame(datos_crudos)

    print("--- DATOS PIZZAS ---")
    ver_esquema(df_pizzas)

except Exception as error:
    print("Error:", error)
```
